```python
def greet(name, language='English'):
    '''say hello to someone in a chosen language.'''
    if language == 'French':
        return f'Bonjour, {name}!'
    return f'Hello, {name}!'


#calling the function
print(greet('Alice'))
print(greet('Bob', 'French'))
```

```python
def greet1(name):
    print(f'hello {name}')
greet1('peter')    
```

## Parameters vs Arguments

```python
def count(list):
    return f'{len(list)}'
list= [1,2,3,4]
print(count(list))
```

```python
def g(a=3):
    return 3+a
g()    
```

```python
def introduction(name,hobby,country):
    message=f'my name is {name} i am from {country} and i love {hobby}'
    return message
```

```python
introduction('Messi', 'football', 'cmr')
```

```python
#key word argument
introduction(hobby='football', country='cmr', name='Messi')
```

```python
# *args : this parameter takes in any number of arguments

def total(*values):
    t=sum(values)
    print(type(t))
    return t
```

```python
total(1,3,4,5,6,7,8,9)
```

```python
# **kwargs : this parameter takes in any number of keyword arguments (e.g a=10)
#            and stores them in a dictionary


def computer_specs(**specs):
    return specs
computer_specs(color='pink', ram_size='16gb')    
```

```python
def computer_specs2(**specs):
    #for i in specs.values(): 
    for i in specs.items():
        print(i)
    return specs    
computer_specs2(x=6,y=6,z=3)
```

## Lambda function

```python
#function to square a number

square=lambda x : x**2
square(5)
```

```python
#sorting using lambda
laliga_table = [('ATM,15'), ('BAR, 22'), ('RMA, 19')]
laliga_table.sort(key=lambda m: m[2], reverse=True)
laliga_table
```

```python
greet=lambda name: f'hello {name}'
greet('john')
```
