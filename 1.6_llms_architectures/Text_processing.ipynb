```python
from sklearn.feature_extraction.text import CountVectorizer

# 1. Create a sample dataset with text reviews
data = {'reviews': ['the food is good',
                           'New food is terrible',
                           'good food and good services']
              }

df = pd.DataFrame(data)

# 2. Initialize the CountVectorizer
# stop_words='english' removes useless words like 'the' and 'and' to save memory
vectorizer = CountVectorizer(stop_words='english')

# 3. Fit and transform the text column 
# This learns the unique vocabulary and counts the word occurrences 
bow_matrix = vectorizer.fit_transform(df['reviews'])

# 4. Convert the matrix into a clean Pandas DataFrame
# vectorizer.get_feature_names_out() extracts the learned vocabulary as column names 

df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())

```

## text cleaning using regular expression

```python
import re 
white_space_splitting = re.split(r'\s', raw_text)
print(white_space_splitting)
```

```python
len(white_space_splitting)
```

```python
raw_text =re.sub(r'@\w+', '',raw_text, flags=re.MULTILINE)
raw_text =re.sub(r'#\w+', '', raw_text, flags=re.MULTILINE)

raw_text = re.sub(r'[^a-zA-Z\s]', '', raw_text)

print(raw_text)
```

```python
#splitting using other criteria
white_space_splitting_with_special_xters = re.split(r'--|[@^!#*?&]|\s', raw_text)
print(white_space_splitting_with_special_xters)
```

```python
#hint : use list comprehension
strip_text = [ word.strip() for word in white_space_splitting_with_special_xters if word.strip()]
len(strip_text)
```

```python
print(strip_text)
```

```python
#sort the text

sorted_strip_text = sorted(set(strip_text))
print(sorted_strip_text)
```

```python
len(sorted_strip_text)
```

```python
#enumerate keeps track 
for idx, token in enumerate(sorted_strip_text):
    print(f'({idx}, {token})')
```

```python
# create vocabulary
vocabulary= {}
for idx, token in enumerate(sorted_strip_text):
    vocabulary[token] = idx
vocabulary    
```

```python
id_to_str = {}
for idx, token in enumerate(vocabulary):
    print(f'({idx}, {token})')
```

```python
id_to_str = {}
for idx, token in enumerate(vocabulary):
    id_to_str[idx] = token
id_to_str     
```

```python
token_ids = [vocabulary[word] for word  in strip_text]
print(token_ids)
```

```python
tokens_recovered = [id_to_str[idx] for idx in token_ids]
' '.join(tokens_recovered)
```
