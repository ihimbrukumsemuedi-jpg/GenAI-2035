# Vectors 
# collection of values 
# an order list of numbers with magnitude

```python
# creating vector in numpy
v=np.array([1,2,3,1])
v
u=np.array([1,2,3,4,5,6])
u
```

```python
#CREATING THE SHAPE OF A VECTOR

print(v.shape)
print(u.shape)
print(v.ndim) # for dimension
print(v.dtype) # for dtype
```

```python
# Vectors Indexing
# Works identically to python list-zero based
v=np.array([10,20,30,40,50])

print(v[0])
print(v[-1])
print(v[:4])
```

## ROWS VS COLUMN IN VECTORS

```python
v=np.array([1,2,5])
print(v.shape)
```

```python
row=v.reshape(1,3)
col=v.reshape(3,1)
print(row.shape)
print(col.shape)
```

# Scalar multiplication

```python
v=np.array([2,4,6])
scalar=2
result=scalar*v
print(result)
```

```python
import matplotlib.pyplot as plt

a=np.array([3,1])
b=np.array([1,3])
c=a+b
print(c)

fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='navy', label='a=[3,1]')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='gold', label='b=[3,1]')
ax.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1, color='red', label='a+b=[4,4')
plt.grid(True)
plt.legend()
#limits of x and y
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
```

# DOT PRODUCT

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

# method 1 explicit formula
manual= a[0]*b[0] + a[1]*b[1] + a[2]*b[2]
print(manual)

# method 2: np.dot (standard)
print(np.dot(a,b))

# c = a+b (normal python without using numpy)
#print c
```

# LENGTH OF A VECTOR

```python
x=np.array([2,3])
print(len(x))
```

```python
v=np.array([3,4])

# manual calculation
manual=np.sqrt(v[0]**2 + v[1]**2)
print(manual)

#numpy 
print(np.linalg.norm(v))    #5.0 (l2 norm, default)
print(np.linalg.norm(v, ord=1))   # 7.0 (l1 norm, [3]+[7])
print(np.linalg.norm(v, ord=np.inf))
```

# A UNIT VECTOR

```python
v=np.array([3.0, 4.0])

unit_v=v / np.linalg.norm(v)
print(unit_v)
print(np.linalg.norm(unit_v))  #confirmed: length is 2
```
