```python
import numpy as np
# we dont have real data: we have to generate data
# note: you are always going to be getting real data from either kaggle or hugging face

np.random.seed(42)  
#1, 2, 4
# 2, 3, 5
# 3, 6, 8
#it might lead to model inconsistency
# seed is going to generate only once and you cab alwaysfit in your model

# we want to create a model that predict the house price given the house size
# generate data for the size
```

```python
size=np.random.randint(50, 201, 20)
price=0.3 * size * np.random.normal(0, 2, 20)
# eg first value of price9=> 0.3*152 *0.5=
print(size)
print(price)
```

```python
# we have to store data in dictionary format:
data={
    'size':size,
    'price':price
}
data
```

```python
# use pandas to convert data into table format: (python library use to manipulated dots in table format)
import pandas as pd
# converting dots into table format is called dataframes
df=pd.DataFrame(data)

df
```
