```python
size=np.random.randint(50, 201, 20)
price=0.3 * size * np.random.normal(0, 2, 20)
# eg first value of price9=> 0.3*152 *0.5=
print(size)
print(price)
```

```python
# we have to store data in dictionary format:
data={
    'size': [50, 60, 70, 80, 90, 100],
    'price':[15, 18, 21, 24, 27, 30]
}
data
```

```python
# use pandas to convert data into table format: (python library use to manipulated dots in table format)
import pandas as pd
# converting dots into table format is called dataframes
df=pd.DataFrame(data)

df
```

```python
# target and features
X=df[['size']]
Y=df[['price']]

X
```

```python
# visualization 
import matplotlib.pyplot as plt
```

```python
# start plotting
plt.scatter(df['size'], df['price'])
plt.xlabel('House size')
plt.ylabel('price in fcfs')
plt.title('House size  vs house price')
plt.show()
```

```python
#start using machine learning library for python


from sklearn.model_selection import train_test_split
```

```python
X_train, X_test,Y_train,Y_test=train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
Y_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#create a model 
model=LinearRegression()
```

```python
# train
model.fit(X_train, Y_train)
```

```python
prediction=model.predict(X_test)
prediction
```

```python
from sklearn.metrics import mean_squared_error

```

```python
mse=mean_squared_error(Y_test, prediction)
print(f' prediction: {prediction}')
print(f" actual:", Y_test.values)
print("MSE", mse)
```
