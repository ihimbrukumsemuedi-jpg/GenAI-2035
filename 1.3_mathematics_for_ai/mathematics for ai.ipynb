```python
#create a function
def  y_square(x):
    return x**2
    
```

```python
y_square(2)
```

```python
x=np.array([1,2,3,4,5])
y_square(x)
```

```python
y=y_square(x)
```

```python
#plot a gragh of y against x
```

```python
#we are going to be using matplotlib
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y, color='navy', linewidth=2)
plt.title('gragh of y against x')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=range(-10, 11)
y=[]
for i  in x:
    y.append(i**3 +1)
plt.plot(x,y)
plt.title('Gragh of y=x^3+1')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

```
