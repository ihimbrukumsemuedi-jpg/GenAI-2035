```python
# HOW TO CREATE A MATRIX IN NUMPY

import numpy as np
A = np.array([[1,2,3,4],
              [5,6,7,8],
              [6,10,11,12]
])
print(A.shape) #3 row and 4 col
print(A.ndim) # dimension
print(A.size) # total elements
print(A.dtype)
```

```python
#slicing element: A[row, col]
print(A[0,0]) #top left corner of first number
print(A[1,2]) #row 1 col 2
print(A[-1, -1]) # bottom left corner

#SILICING AN entire row
print(A[1, :])
print(A[1])
```

```python
import numpy as np
import matplotlib.pyplot as plt
m=np.random.rand(5, 5)
plt.figure(figsize=(6,5))
#cmar= np.random.rand(3, 3) # generates number btw 0 and 1
#plt.imshow(m, cmar)
#print(r)p='Blues'
#plt.colorbar()
#plt.title('Matrix as a Heatmap')
plt.imshow(m cmap='Blues')  #'cmap' sets the color scheme
plt.colorbar()              # show the color scale
plt.title('Matrix as a Heatmap')


# Annotate each cell with its value
for i in range(5):
    for j in range(5):
        plt.text(j, i, f'{m[i,j]:.2f}',
                ha='center', va='center', fontsize=9)
plt.show()        
```
